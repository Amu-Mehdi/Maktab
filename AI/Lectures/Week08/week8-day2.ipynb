# Binary Cross-Entropy Loss (BCE Loss)

## I. Introduction

**Binary Cross-Entropy (BCE)**, also called **Log Loss**, is the most common loss function used for **binary classification problems**.

It measures the difference between:

- The **true label** $y$
- The **predicted probability** $\hat{y}$

The goal of training is to minimize this difference.

---

# II. Binary Classification Output

In binary classification, the target variable has two possible values:

```
y = 0 → Negative class

y = 1 → Positive class
```

A neural network usually produces a probability using a **sigmoid activation function**:

$$
\hat{y}=\sigma(z)
$$

where:

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

The output is always between 0 and 1:

```
0.0 ───────────────── 1.0

Low probability       High probability
```

Example:

```
Prediction = 0.92

Meaning:
The model is 92% confident that the sample belongs to class 1.
```

---

# III. Binary Cross-Entropy Formula

For one training example:

$$
L(y,\hat{y})
=
-\left[
y\log(\hat{y})
+
(1-y)\log(1-\hat{y})
\right]
$$

where:

| Symbol | Meaning |
|---|---|
| $y$ | True label (0 or 1) |
| $\hat{y}$ | Predicted probability |
| $L$ | Loss value |

---

# IV. Understanding the Formula

The formula has two parts:

$$
L
=
-
[
\underbrace{y\log(\hat{y})}_{\text{Class 1 term}}
+
\underbrace{(1-y)\log(1-\hat{y})}_{\text{Class 0 term}}
]
$$

Only one term contributes because $y$ is either 0 or 1.

---

# V. Case 1: True Label is 1

If:

$$
y=1
$$

The formula becomes:

$$
L
=
-\log(\hat{y})
$$

The loss depends only on the probability assigned to class 1.

---

## Example: Correct Prediction

True label:

$$
y=1
$$

Prediction:

$$
\hat{y}=0.95
$$

Loss:

$$
L=-\log(0.95)
$$

$$
L=0.051
$$

Small loss.

The model is confident and correct.

---

## Example: Wrong Prediction

True label:

$$
y=1
$$

Prediction:

$$
\hat{y}=0.05
$$

Loss:

$$
L=-\log(0.05)
$$

$$
L=2.996
$$

Large loss.

The model is confident but wrong.

---

# VI. Case 2: True Label is 0

If:

$$
y=0
$$

The formula becomes:

$$
L
=
-\log(1-\hat{y})
$$

The loss depends on the probability assigned to class 0.

---

## Example: Correct Prediction

True label:

$$
y=0
$$

Prediction:

$$
\hat{y}=0.10
$$

Loss:

$$
L=-\log(1-0.10)
$$

$$
L=0.105
$$

Small loss.

---

## Example: Wrong Prediction

True label:

$$
y=0
$$

Prediction:

$$
\hat{y}=0.90
$$

Loss:

$$
L=-\log(1-0.90)
$$

$$
L=2.303
$$

Large loss.

---

# VII. Loss Behavior Intuition

The logarithm creates a strong penalty for confident wrong predictions.

```
Prediction Probability

0        0.5              1
|---------|---------------|

For y = 1:

Low probability → Large loss

High probability → Small loss


For y = 0:

Low probability → Small loss

High probability → Large loss
```

---

# VIII. Dataset Loss (Average BCE)

For a dataset with $N$ samples:

$$
J(\theta)
=
-\frac{1}{N}
\sum_{i=1}^{N}
\left[
y_i\log(\hat{y}_i)
+
(1-y_i)\log(1-\hat{y}_i)
\right]
$$

where:

- $N$ = number of training examples
- $\theta$ = model parameters (weights and biases)

The optimizer tries to minimize:

$$
\min_\theta J(\theta)
$$

---

# IX. Example in Neural Network Training

Suppose a neural network predicts:

| Sample | True Label | Prediction |
|---|---|---|
| 1 | 1 | 0.90 |
| 2 | 0 | 0.20 |
| 3 | 1 | 0.60 |

For each sample:

```
Sample 1:
Correct positive prediction
→ Small loss


Sample 2:
Correct negative prediction
→ Small loss


Sample 3:
Weak positive prediction
→ Larger loss
```

The total BCE loss is the average of these errors.

---

# X. Why Not Use Mean Squared Error?

For binary classification, we could use:

$$
MSE=(y-\hat{y})^2
$$

but BCE is preferred because:

| MSE | Binary Cross-Entropy |
|---|---|
| Treats classification as regression | Designed for probability prediction |
| Smaller gradients near wrong confident predictions | Strong penalty for confident mistakes |
| Can create non-convex optimization with sigmoid | Convex for logistic regression |
| Slower learning | Faster convergence |

---

# XI. BCE and Sigmoid Together

The common architecture is:

```
Input Features

      |
      v

Linear Layer

z = Wx + b

      |
      v

Sigmoid Activation

ŷ = σ(z)

      |
      v

Binary Cross-Entropy Loss

L(y, ŷ)
```

The sigmoid converts the output into a probability.

The BCE loss evaluates whether that probability is correct.

---

# XII. Gradient of BCE Loss

A major advantage of BCE with sigmoid is the simple gradient:

$$
\frac{\partial L}{\partial z}
=
\hat{y}-y
$$

This means:

```
Gradient = Prediction - Actual Value
```

Example:

```
Prediction = 0.8
Actual     = 1

Gradient = 0.8 - 1

Gradient = -0.2
```

The negative gradient tells the optimizer:

> Increase the prediction.

---

# XIII. PyTorch Implementation

Using PyTorch:

`````python
import torch
import torch.nn as nn

loss_function = nn.BCELoss()

prediction = torch.tensor(
    [[0.9]]
)

target = torch.tensor(
    [[1.0]]
)

loss = loss_function(
    prediction,
    target
)

print(loss)
`````


---

# XIV. More Stable Implementation

In practice, PyTorch usually combines:

- Sigmoid
- Binary Cross-Entropy

into one function:

```python
loss_function = nn.BCEWithLogitsLoss()
```

Instead of:

```
Linear Layer
      |
      v
Sigmoid
      |
      v
BCELoss
```

we use:

```
Linear Layer
      |
      v
BCEWithLogitsLoss
```

Advantages:

- Better numerical stability
- Avoids floating-point problems
- Recommended for neural networks

# Key Takeaway

**Binary Cross-Entropy measures how close the predicted probability is to the true binary label.**

A model is rewarded when:

$$
\hat{y}\approx y
$$

and strongly penalized when it is:

- confident
- but wrong.

For binary classification neural networks:

$$
\boxed{
\text{Linear Layer}
+
\text{Sigmoid}
+
\text{Binary Cross-Entropy}
}
$$

forms the standard learning framework.

# 5. Optimization Problem

Training a neural network means solving:

> Find weights that minimize the loss function.

Mathematically:

$$
minimize \ L(W)
$$


where:

- W = model weights
- L = loss function

---

# 6. Gradient Descent

Gradient Descent is the algorithm used to update weights.

The idea:

> Move toward the direction where the loss decreases.

Imagine a mountain:

```

```
    Loss

      ▲

      |

   ● Starting Point


      ↓


   Lowest Point
```

```

The lowest point represents:

Minimum error.

---

# Mathematical Formulation

Weight update:

$$
\theta_{new}
=
\theta_{old}
-
\alpha\nabla L(\theta)
$$


Where:

| Symbol | Meaning |
|-|-|
| θ | Model parameters |
| α | Learning rate |
| ∇L | Gradient of loss |

# 8. Gradient Descent Variants

## Batch Gradient Descent

Uses all training examples.

Advantages:

- Accurate gradient

Disadvantages:

- Slow for large datasets


---

## Stochastic Gradient Descent (SGD)

Uses one sample at a time.

Advantages:

- Faster

Disadvantages:

- Noisy updates


---

## Mini-Batch Gradient Descent

Uses a small group of samples.

Example:

```

Batch size = 32

```

This is the most common approach in modern Deep Learning.

# 4. Creating a Model in PyTorch

```python
import torch
import torch.nn as nn


class SimpleNN(nn.Module):

    def __init__(self):

        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(4,8),

            nn.ReLU(),

            nn.Linear(8,1),

            nn.Sigmoid()

        )


    def forward(self,x):

        return self.network(x)
```

# 8. Training Loop

The complete learning process:

```python
for epoch in range(100):

    # Forward pass
    prediction = model(X)

    # Calculate loss
    loss = criterion(prediction,y)


    # Clear old gradients
    optimizer.zero_grad()


    # Backpropagation
    loss.backward()


    # Update weights
    optimizer.step()


    print(loss.item())
```
