```python
# Problem 1:
import pandas as pd

df = pd.read_csv("student_performance_session2.csv")

#1. Display the first five rows of the dataset.
print("===== First 5 rows of the dataset =====")
print(df.head())
print("\n" + "="*50 + "\n")

#2. Display the general information about the dataset.
print("===== General information about the dataset (info) =====")
print(df.info())
print("\n" + "="*50 + "\n")

#3. Display the descriptive statistics of the numerical columns.
print("===== Descriptive statistics of numerical columns =====")
print(df.describe())
print("\n" + "="*50 + "\n")

#4. Based on the output of `info()`, state the number of rows and columns in the dataset.
num_rows, num_cols = df.shape
print(f"===== Number of rows and columns =====")
print(f"Number of rows: {num_rows}")
print(f"Number of columns: {num_cols}")
print("\n" + "="*50 + "\n")

#5. Identify the columns whose data types are numerical.
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
print("===== Numerical columns =====")
print(f"Count: {len(numeric_cols)}")
print(f"Column names: {numeric_cols}")
print("\n" + "="*50 + "\n")

#6. Identify the columns whose data types are `object`.
object_cols = df.select_dtypes(include=['object']).columns.tolist()
print("===== Object columns =====")
print(f"Count: {len(object_cols)}")
print(f"Column names: {object_cols}")
print("\n" + "="*50 + "\n")
```

```python
#Problem 3:
import pandas as pd

# Load dataset
df = pd.read_csv('online_orders_classwork.csv')

# 1. Number of orders in each city
city_counts = df['City'].value_counts()
print("Orders by City:")
print(city_counts)

# 2. Number of orders in each product category
category_counts = df['ProductCategory'].value_counts()
print("\nOrders by Product Category:")
print(category_counts)

# 3. Frequency of each payment method
payment_counts = df['PaymentMethod'].value_counts()
print("\nFrequency of Payment Methods:")
print(payment_counts)

# 4. Number of returned and non-returned orders
returned_counts = df['Returned'].value_counts()
print("\nReturn Status:")
print(returned_counts)

# Answer questions
print("\n" + "="*50)
print("ANSWERS:")
print("="*50)

# Question 1
print(f"\n1. City with highest orders: {city_counts.index[0]} ({city_counts.iloc[0]} orders)")

# Question 2
print(f"2. Category with lowest orders: {category_counts.index[-1]} ({category_counts.iloc[-1]} orders)")

# Question 3
print(f"3. Most used payment method: {payment_counts.index[0]} ({payment_counts.iloc[0]} orders)")

# Question 4
# Assuming 'Yes' or True represents returned orders
returned_count = returned_counts.get('Yes', returned_counts.get(True, 0))
print(f"4. Number of returned orders: {returned_count}")

# Question 5
non_returned_count = returned_counts.get('No', returned_counts.get(False, 0))
if non_returned_count > returned_count:
    print(f"5. Yes, non-returned orders ({non_returned_count}) are more than returned orders ({returned_count})")
else:
    print(f"5. No, returned orders ({returned_count}) are more than non-returned orders ({non_returned_count})")
```
