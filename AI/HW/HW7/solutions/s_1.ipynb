### Part B:
#### a)

Scenario 6 has **no ground-truth labels** to compare predictions against — Accuracy and F1-score both require knowing the "correct" class for each sample. In clustering, there's no true answer to check the cluster assignments against, so these metrics simply can't be computed. (Clustering quality is instead judged with metrics like Silhouette Score, which measure internal structure, not label agreement.)

---

#### b)

Since only a small fraction of transactions are actually fraudulent, a model that predicts **"legitimate" for everything** would still achieve very high Accuracy — it would just be missing almost all the fraud cases. High Accuracy here mostly reflects the class imbalance, not real predictive skill, so Precision, Recall, and F1-score are needed to see how well the model actually catches fraud.

---

#### c)

A Regression model (Scenario 1) outputs a **continuous number** (e.g., minutes watched), so we evaluate it by how close predictions are to actual values (MAE, MSE, R²). A Classification model (Scenario 2) outputs a **discrete category** (Positive/Negative), so evaluation is based on how many predictions matched the correct class (Accuracy, Precision/Recall/F1). Because the output types are fundamentally different (numeric vs. categorical), each needs its own family of metrics — distance-based for regression, agreement-based for classification.
